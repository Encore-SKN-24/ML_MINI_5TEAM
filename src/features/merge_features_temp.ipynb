### data 합치는 로직

```python
import pandas as pd

def load_and_merge_data():
    # dict 선언: key는 데이터셋 이름, value는 path와 features를 가진 dict
    file_dict = {
        "rain": {"path": "data/rain.csv", "features": ["DAM_DAY_RNFL", "DATE"]},
        "river": {"path": "data/river.csv", "features": ["RIVER_SE", "DISTANCE"]},
        "soil": {"path": "data/soil.csv", "features": ["SOILDRA", "AREA"]},
        "area": {"path": "./data/processed/merged.csv", "feature": ["CTPRVN_CD","CTPRVN_NAME","SGG_CD","SGG_NAME","AREA_NAME",]}
    }

    # 최종 완성한 csv 파일 저장할 경로
    output_path = "./data/processed/merged.csv"

    # 최종 완성할 df
    merged_df = None

    # 공통 변수 이름
    SIDO, SIGUNGU = "CTPRVN_CD", "SGG_CD"

    # 각 파일을 불러와서 필요한 변수만 추출 후 병합
    for key, file_item in file_dict.items():
        file_path = file_item["path"]
        file_features = file_item["features"]

        # 기본적으로 CTPRVN_CD, SGG_CD는 항상 포함
        cols_to_use = [SIDO, SIGUNGU] + file_features
        df = pd.read_csv(file_path, usecols=cols_to_use)

        # 최초 실행 때는 df 가 null인 상태라 초기화 넣어줌
        if merged_df is None:
            merged_df = df
        else:
            # CTPRVN_CD, SGG_CD 기준으로 병합
            merged_df = pd.merge(merged_df, df, on=[SIDO, SIGUNGU], how="outer")

    merged_df.to_csv(output_path)
    return merged_df
```
