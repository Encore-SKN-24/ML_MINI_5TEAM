```python
import pandas as pd
import numpy as np
import io
```

```python
df_sido = pd.read_csv("./data/raw/sido.txt", 
                 header=None, 
                 names=['CTPRVN_CD', 'CTPRVN_NAME'], 
                 quotechar="'", 
                 skipinitialspace=True)
df_sido.head()
```

```python
df_sigungu = pd.read_csv("./data/raw/sigungu.txt", 
                 header=None, 
                 names=['SGG_CD', 'CTPRVN_CD', 'SGG_NAME'], 
                 quotechar="'", 
                 skipinitialspace=True)
df_sigungu.head()
```

```python
df_merged = pd.merge(df_sido, df_sigungu, on='CTPRVN_CD', how='inner')
df_merged['AREA_NAME'] = df_merged['CTPRVN_NAME'] + " " + df_merged['SGG_NAME'] 
df_merged.to_csv("./data/processed/area.csv", index=False)
df_merged.head(20)
```
