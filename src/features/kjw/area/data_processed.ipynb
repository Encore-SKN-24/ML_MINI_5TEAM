```python
flood_area_df=pd.read_csv('C:/SKN_24/ML_MINI_5TEAM/data/processed/flood_area.csv')
flood_dis_df=pd.read_csv('C:/SKN_24/ML_MINI_5TEAM/data/processed/Flooding_Disaster.csv')
riv_df=pd.read_csv('C:/SKN_24/ML_MINI_5TEAM/data/processed/riv_pro_df.csv')
solid_df=pd.read_csv('C:/SKN_24/ML_MINI_5TEAM/data/processed/solid_pump_data.csv')
waether_df=pd.read_csv('C:/SKN_24/ML_MINI_5TEAM/data/processed/weather_rain_data.csv')
```
