# 행정동 코드로 변경

```python
import pandas as pd
import numpy as np
```

1. 시군구 경계 전처리 파일(pd.read)

```python
# sigungu_pro(시군구 경계 전처리) 파일 -> 팀원과 약속한 행정동 기준 지역코드로 통합하기 위한 파일 불러오기
df = pd.read_csv('../data/sigungu_pro.csv', encoding= 'cp949')
df
```

```python
df.info()
```

```python
# merge 시키기 위해 'SIGUNGU_CD'도 남겨두기
df=df[['SIGUNGU_CD','지역코드']]
df
```

```python
type(df)
```

```python
df.info()
df.describe()
```

---

2. 하천 규모 전처리 파일 (pd.read)

```python
# river_pro 파일 
#   - RIV_DIS_MIN.csv 파일과 국가기본도_하천중심선_전처리.csv 파일 merge한 전처리 파일
#       - 깃허브에 올릴 때는 위 두 파일 없음. 전처리 결과 csv 파일만 올릴 예정. 폴더 효율성을 위함
#   - 하천 규모 5등급 -> 3등급으로 전처리도 함
#   - 즉, 하천 관련 정보는 다 있으나 시군구 코드 등 지역코드가 아직 통일이 안된 상태
riv_df=pd.read_csv('../data/river_pro.csv')
riv_df
```

```python
riv_df.info()
riv_df.describe()
```

---

# MERGE 2

```python
merged_df = pd.merge(riv_df,df,on='SIGUNGU_CD',how='inner')
```

```python
merged_df
```

```python
# 필요 없는 컬럼 제외 
merged_df=merged_df.drop(columns=['RIVER_SE','SIGUNGU_CD','SIGUNGU_NM'])
merged_df
```

```python
# 지역코드 컬럼명 변경 -> SIGUNGU
merged_df=merged_df.rename(columns={'지역코드': 'SIGUNGU'})
merged_df
```

---

# CSV 저장

```python
riv_pro_df = merged_df.copy()

#  CSV 저장
riv_pro_df.to_csv("../data/riv_pro_df.csv", index=False, encoding="utf-8-sig")
print(" 저장 완료: ../data/riv_pro_df.csv")
```
