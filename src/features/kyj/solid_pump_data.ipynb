```python
import pandas as pd
```

```python
df_1 = pd.read_csv('./data/pump_all_data.csv', encoding='cp949')
# df_1 = df_1.filter(items=['시설명', '시도명', '시군구명', '위도', '경도', '제공기관코드', '제공기관명'])

df_1
```

```python
# 예외 처리
df_1.loc[df_1['소재지지번주소'].str.contains('덕양구', na=False), '시군구명'] = '고양시 덕양구'
df_1.loc[df_1['소재지지번주소'].str.contains('일산서구', na=False), '시군구명'] = '고양시 일산서구'
df_1.loc[df_1['소재지지번주소'].str.contains('오정구', na=False), '시군구명'] = '부천시 오정구'
df_1.loc[df_1['소재지지번주소'].str.contains('소사구', na=False), '시군구명'] = '부천시 소사구'
df_1.loc[df_1['소재지지번주소'].str.contains('영통구', na=False), '시군구명'] = '수원시 영통구'
df_1.loc[df_1['소재지지번주소'].str.contains('권선구', na=False), '시군구명'] = '수원시 권선구'
df_1.loc[df_1['소재지지번주소'].str.contains('만안구', na=False), '시군구명'] = '안양시 만안구'
df_1.loc[df_1['소재지지번주소'].str.contains('동안구', na=False), '시군구명'] = '안양시 동안구'

df_1
```

```python
df_1.info()
df_1.describe()
```

```python
df_1 = df_1.rename(columns = {'시도명': 'SIDO_N', '시군구명': 'SIGUNGU_N', '시설명': 'PUMP_CNT'})
df_1
```

```python
# 시도명, 시군구명으로 집계
groupby_df_1 = df_1.groupby(['SIDO_N', 'SIGUNGU_N'])
groupby_df_1.count()
```

```python
# 필요한 컬럼 값만 남김
df_1 = groupby_df_1.count()
df_1 = df_1.filter(items=['SIDO_N', 'SIGUNGU_N', 'PUMP_CNT'])
df_1
```

```python
# 데이터 평면화
df_1 = df_1.reset_index()
df_1
```

```python
# 해당 DF에 지역코드 매칭을 위해 새로운 컬럼 생성
df_1['SIGUNGU_FNM'] = df_1['SIDO_N'] + ' ' + df_1['SIGUNGU_N'] 
df_1
```

```python
df_2 = pd.read_csv('./data/solid_data.csv', encoding='cp949')
df_2
```

```python
all_data = pd.merge(df_1, df_2, on = 'SIGUNGU_FNM', how = 'left')

all_data
```

```python
all_data = all_data.filter(items=['SIGUNGU_CD', 'CODE_DC', 'PUMP_CNT'])
all_data = all_data.rename(columns = {'SIGUNGU_CD': 'SIGUNGU', 'CODE_DC': 'DRAINAGE_GRD'})
all_data
```
